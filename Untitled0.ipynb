```python
import yfinance as yf
import pandas as pd
from datetime import datetime

# 設定股票代號（台積電在台股代號為 2330.TW）
ticker = "2330.TW"

# 設定時間範圍自 2026-01-01 開始至今
start_date = "2026-01-01"
end_date = datetime.today().strftime('%Y-%m-%d')

# 抓取歷史數據
df = yf.download(ticker, start=start_date, end=end_date)

# 整理欄位名稱並顯示前幾筆資料
if not df.empty:
    display(df.head())
else:
    print("未找到 2026 年起的歷史數據，請確認當前年份與市場交易日。")
```

```python
import plotly.graph_objects as gp
from plotly.subplots import make_subplots

if not df.empty:
    # 建立雙 Y 軸圖表：上方為 K 線圖，下方為成交量
    fig = make_subplots(rows=2, cols=1, shared_xaxes=True,
                        vertical_spacing=0.1,
                        subplot_titles=('台積電 (2330) K線圖', '成交量 (Volume)'),
                        row_width=[0.3, 0.7])

    # 1. K線圖 (Candlestick)
    fig.add_trace(gp.Candlestick(
        x=df.index,
        open=df['Open'],
        high=df['High'],
        low=df['Low'],
        close=df['Close'],
        name='K線'
    ), row=1, col=1)

    # 2. 成交量圖 (Bar)
    fig.add_trace(gp.Bar(
        x=df.index,
        y=df['Volume'],
        name='成交量',
        marker_color='blue'
    ), row=2, col=1)

    # 調整外觀設定
    fig.update_layout(
        title='台積電 (2330) 2026年至今股價走勢圖',
        yaxis_title='價格 (TWD)',
        yaxis2_title='股數 (Shares)',
        xaxis_rangeslider_visible=False, # 隱藏下方的滑桿
        height=800,
        template='plotly_white'
    )

    fig.show()
else:
    print("無資料可繪製圖表。")
```

```python
import yfinance as yf
import pandas as pd
from datetime import datetime

# 設定股票代號與新的時間範圍（2024年初至今）
ticker = "2330.TW"
start_date = "2024-01-01"
end_date = datetime.today().strftime('%Y-%m-%d')

# 重新下載資料
df_2024 = yf.download(ticker, start=start_date, end=end_date)

# 處理 MultiIndex 欄位結構（若 yfinance 返回多級欄位則取第一級）
if isinstance(df_2024.columns, pd.MultiIndex):
    df_2024.columns = df_2024.columns.droplevel(1)

# 計算 5日、20日、60日移動平均線
df_2024['MA5'] = df_2024['Close'].rolling(window=5).mean()
df_2024['MA20'] = df_2024['Close'].rolling(window=20).mean()
df_2024['MA60'] = df_2024['Close'].rolling(window=60).mean()

# 顯示前幾筆與後幾筆資料確保計算正確
display(df_2024.tail())
```

```python
import plotly.graph_objects as gp
from plotly.subplots import make_subplots

if not df_2024.empty:
    # 建立雙軸子圖：上方為 K 線 + 均線，下方為成交量
    fig = make_subplots(rows=2, cols=1, shared_xaxes=True,
                        vertical_spacing=0.1,
                        subplot_titles=('台積電 (2330) K線與移動平均線 (2024年至今)', '成交量 (Volume)'),
                        row_width=[0.3, 0.7])

    # 1. K線圖
    fig.add_trace(gp.Candlestick(
        x=df_2024.index,
        open=df_2024['Open'],
        high=df_2024['High'],
        low=df_2024['Low'],
        close=df_2024['Close'],
        name='K線'
    ), row=1, col=1)

    # 2. 5日均線 (MA5)
    fig.add_trace(gp.Scatter(
        x=df_2024.index, y=df_2024['MA5'],
        mode='lines',
        name='5日均線 (MA5)',
        line=dict(color='orange', width=1.5)
    ), row=1, col=1)

    # 3. 20日均線 (MA20)
    fig.add_trace(gp.Scatter(
        x=df_2024.index, y=df_2024['MA20'],
        mode='lines',
        name='20日均線 (MA20)',
        line=dict(color='magenta', width=1.5)
    ), row=1, col=1)

    # 4. 60日均線 (MA60)
    fig.add_trace(gp.Scatter(
        x=df_2024.index, y=df_2024['MA60'],
        mode='lines',
        name='60日均線 (MA60)',
        line=dict(color='green', width=1.5)
    ), row=1, col=1)

    # 5. 成交量圖
    fig.add_trace(gp.Bar(
        x=df_2024.index,
        y=df_2024['Volume'],
        name='成交量',
        marker_color='blue'
    ), row=2, col=1)

    # 版面樣式調整
    fig.update_layout(
        title='台積電 (2330) 2024年至今股價走勢與移動平均線',
        yaxis_title='價格 (TWD)',
        yaxis2_title='股數 (Shares)',
        xaxis_rangeslider_visible=False,
        height=800,
        template='plotly_white'
    )

    fig.show()
else:
    print("無法取得資料以繪製圖表。")
```

```python
import pandas as pd

# 確保 df_2024 的索引為 DatetimeIndex
df_2024.index = pd.to_datetime(df_2024.index)

# 1. 取得每個月底最後一筆交易日的收盤價，以及每個月初第一筆交易日的開盤價
monthly_data = df_2024.resample('ME').agg({
    'Open': 'first',
    'Close': 'last'
})

# 2. 計算每個月的漲跌絕對值與漲幅百分比
monthly_data['Monthly_Gain_TWD'] = monthly_data['Close'] - monthly_data['Open']
monthly_data['Monthly_Return_Pct'] = (monthly_data['Monthly_Gain_TWD'] / monthly_data['Open']) * 100

# 3. 排序找出漲幅最大（最凶猛）的月份
monthly_gain_sorted = monthly_data.sort_values(by='Monthly_Return_Pct', ascending=False)

print("--- 按單月漲幅(%)排序的前 5 個月份 ---")
display(monthly_gain_sorted[['Open', 'Close', 'Monthly_Gain_TWD', 'Monthly_Return_Pct']].head(5))
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 讀取 Iris.csv 資料集
iris_df = pd.read_csv('/content/Iris.csv')

# 顯示資料集的前五筆，確認欄位名稱
display(iris_df.head())
```

```python
# 如果資料集中有 'Id' 欄位，在畫圖前先將其排除以免干擾視覺化
if 'Id' in iris_df.columns:
    plot_df = iris_df.drop(columns=['Id'])
else:
    plot_df = iris_df.copy()

# 繪製 Pairwise Scatter Plot
sns.set_theme(style="ticks")
pair_plot = sns.pairplot(plot_df, hue="Species", palette="husl", markers=["o", "s", "D"])

# 設定圖表標題
pair_plot.fig.suptitle("Iris Dataset Pairwise Scatter Plot", y=1.02, fontsize=16)

# 顯示圖表
plt.show()
```

```python
# 排除 'Id' 欄位後，依品種 (Species) 分組並計算平均值
features_mean = iris_df.drop(columns=['Id'], errors='ignore').groupby('Species').mean()

# 顯示計算結果
display(features_mean)
```
